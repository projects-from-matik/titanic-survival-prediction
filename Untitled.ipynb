```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
url = 'https://raw.githubusercontent.com/tflearn/tflearn.github.io/master/resources/titanic_dataset.csv'
ds = pd.read_csv(url)
```

```python
ds.columns
```

```python
labels = pd.get_dummies(ds.iloc[:, 0]).to_numpy(dtype=np.float32)
data = ds.iloc[:, 1:].values.tolist()
```

```python
def preprocess(data, discarded_columns):
    for id in sorted(discarded_columns, reverse=True):
        [r.pop(id) for r in data]
        
    for i in range(len(data)):
        data[i][1] = 1. if data[i][1] == 'female' else 0.
        
    return np.array(data, dtype=np.float32)
```

```python
discarded_columns = [1, 6]
data = preprocess(data, discarded_columns)
```

```python
data
```

```python
model = models.Sequential([
    layers.Input(shape=(6,)),                   
    layers.Dense(32, activation='relu'), 
    layers.Dense(32, activation='relu'),        
    layers.Dense(2, activation='softmax')   
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(data, labels, epochs=20, batch_size=16)
```

```python
Jack = [3, 'Jack London', 'male', 19, 0, 0, 'N/A', 5.0000]
Rose = [1, 'Rose Key', 'female', 17, 1, 2, 'N/A', 100.0000]

dicarded_columns = [1, 6]
passengers = preprocess([Jack, Rose], dicarded_columns)
```

```python
pred = model.predict(passengers)

pred
```
